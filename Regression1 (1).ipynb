```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#1.)
dataset = pd.read_csv("/Users/rausch/Desktop/Python/ai/data - Sheet1.csv")
x = dataset.iloc[:, 0:1].values
y = dataset.iloc[:, -1].values
```

```python
#2.)
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 1/3, random_state = 0)
```

```python
#3.) SLR with hours studied
from sklearn import metrics
#creating regressor
regressor = LinearRegression()
#fitting
regressor.fit(x_train,y_train)
#finding y_predict
y_predict = regressor.predict(x_test)
MSE = metrics.mean_squared_error(y_test, y_predict)
plt.scatter(x_train, y_train, color="Red")
plt.plot(x_train, regressor.predict(x_train), color="Blue")
print(MSE)
```

```python
#5.)
xmulti = dataset.iloc[:,:-1].values
xmulti_train, xmulti_test, ymulti_train, ymulti_test = train_test_split(xmulti,y,test_size = 1/3, random_state = 0)
regressormulti = LinearRegression()
regressormulti.fit(xmulti_train, ymulti_train)
ymulti_predict = regressormulti.predict(xmulti_test)
MSEmulti = metrics.mean_squared_error(ymulti_test, ymulti_predict)
print(MSEmulti)
```

```python
#5.)
import statsmodels.api as sm
MSEmulti = metrics.mean_squared_error(ymulti_test, ymulti_predict)
model = sm.OLS(y_train, x_train).fit()
multimodel = sm.OLS(ymulti_train, xmulti_train).fit()
print("this is the multimodel summary \n", multimodel.summary(), "\n")
print( "this is the single model summary \n", model.summary())
if MSE > MSEmulti:
    print("MSE is higher for simple linear than multi linear, therefore the Multiple linear regression function is more accurate")
else:
    print("MSE is higher for multi linear than simple linear, therefore the simple linear regression function is more accurate")
    
```
